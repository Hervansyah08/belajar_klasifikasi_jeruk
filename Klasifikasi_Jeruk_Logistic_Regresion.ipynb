```python
from logging import warn
from sklearn.linear_model import LogisticRegression # Walaupun namanya regression, Logistic Regression itu model klasifikasi
from sklearn.model_selection import train_test_split # membagi data menjadi data latih dan data test
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
# StandardScaler Agar komputer tidak menganggap nilai yang besar itu lebih penting / lebih baik hanya karena angkanya besar, jadi kayak menggangap setara
# mengepress angka nya agar sama seperti yang lain

# OneHotEncoder Untuk data kategori tanpa urutan.
# contoh merah, kuning , hijau. ini kan tidak ada urutannya (merah bukan lebih besar dari kuning), makanmya OneHotEncoder = mengubah kategori menjadi kolom biner (0 & 1)
# agar komputer menganggap Tidak ada angka yang “lebih besar”

# OrdinalEncoder Untuk data kategori yang ada urutan, OrdinalEncoder = mengubah kategori berurutan jadi angka berurutan
# contoh jelek  → 0 sedang → 1 bagus  → 2

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.pipeline import Pipeline # Menggabungkan preprocessing + model jadi satu alur. Tanpa Pipeline: Scaling manual, Encoding manual, Model manual
# jadi nanti output nya 1 model saja

from sklearn.compose import ColumnTransformer  # Mengatur kolom mana pakai preprocessing apa. Contoh: Kolom angka → StandardScaler ,Kolom kategori → OneHotEncoder

x = df[["diameter", "berat", "tebal_kulit", "kadar_gula", "asal_daerah", "warna", "musim_panen"]]
y = df["kualitas"]

# pembagian dataset train dan test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# data preprosesing
numeric_columns = ["diameter", "berat", "tebal_kulit", "kadar_gula"]
categorical_columns = ["asal_daerah", "musim_panen"]
ordinal_columns = ["warna"]
# karena warna ini ada urutannya dari yang jelek sampai bagus
# maka kita perlu menyiapkan urutannya
warna_order = ["hijau", "kuning", "oranye"]
ordinal_order = [warna_order]

# ini preprocessing di lakukan di satu ruangan, harusnya aslinya di pisah pisah
prerocessing = ColumnTransformer(
    transformers=[
        ("scaler", StandardScaler(), numeric_columns),
        ("ohe", OneHotEncoder(), categorical_columns),
        ("oe", OrdinalEncoder(categories=ordinal_order), ordinal_columns)
    ]
)

model = Pipeline(
    steps=[
        ("preprocessor", prerocessing),
        ("classifier", LogisticRegression())
    ]
)


model.fit(x_train, y_train) # ini perintah agar model belajar menggunakan data latih
y_pred = model.predict(x_test) # ini perintah agar model ulangan menggunakan data tes, yang disimpan dalam variable y_pred

# ini perintah untuk tes akurasi
print("Akurasi: ", accuracy_score(y_test, y_pred)) # kita mencocokan jawaban ulangan yaitu y_pred dengan kunci jawaban ulangan yaitu y_test

print("\nClassification Report:\n", classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

Classification Report

precision (ketepatan prediksi). kalau hasinya 1.00, berarti Semua yang

diprediksi benar sesuai kunci jawaban

jadi “Kalau model bilang ini Bagus, bener nggak?”


---



recall (ketangkapan data asli)

“Dari semua jeruk Bagus yang ada, ketangkep semua nggak?”



---

f1-score (nilai keseimbangan)

Gabungan precision + recall

Kalau f1-score tinggi:
Model imbang,
Tidak asal tebak


---




```python
# coba prediksi
data_baru = pd.DataFrame([[7.90, 200, 0.40, 10, "Kalimantan", "oranye", "kemarau"]],
                         columns=["diameter", "berat", "tebal_kulit", "kadar_gula", "asal_daerah", "warna", "musim_panen"])
prediksi = model.predict(data_baru)[0]
persentase = max(model.predict_proba(data_baru)[0])
print(model.predict_proba(data_baru)[0]) # perintah untuk melihat probabilitas, jadi nanti urutannya Bagus, Jelek ,Sedang
print(f"Model memprediksi {prediksi} dengan tingkat keyakinan {persentase*100:.2f}%")
```
